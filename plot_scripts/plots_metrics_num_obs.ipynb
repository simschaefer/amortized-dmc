```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import pickle
import seaborn as sns
import os

network_name = 'updated_priors_sdr_estimated'
parent_dir = os.path.dirname(os.getcwd())

# Load model specification
model_specs_path = parent_dir + '/model_specs/model_specs_' + network_name + '.pickle'

with open(model_specs_path, 'rb') as file:
    model_specs = pickle.load(file)

data_path = parent_dir + '/data_complete/insilico/' + network_name

network_plot_folder = parent_dir + '/plots/metrics_num_obs/' + network_name 

os.makedirs(network_plot_folder, exist_ok=True)

num_data_sets = 100
```

```python

fontsize = 22

data_set_metrics = pd.read_csv(data_path + '/' + network_name + '_metrics_num_obs.csv')

fig, axes = plt.subplots(1,len(model_specs['simulation_settings']['param_names']),
                         sharey=True, sharex=True, 
                         gridspec_kw=dict(left=None, bottom=None, right=None, top=None, wspace=0.0, hspace=None),
                        figsize=(18.5,3.0))

hue_order = ["Calibration Error", "Contraction Factor", "NRMSE"]
palette = {"Calibration Error": "#8a90a0", "Contraction Factor": "#f7a85e", "NRMSE": "#132a70"}


for p, ax in zip(model_specs['simulation_settings']['param_names'], axes):
    
    suff = "$\\" if p in ["tau", "mu_c", "mu_r"] else "$"

    label = suff + p + "$"

    sns.lineplot(data_set_metrics[data_set_metrics["variable_names"] == p], x="num_obs_bin", y="values", hue="metric_name", ax=ax, hue_order=hue_order, palette=palette, errorbar='sd')
    sns.scatterplot(data_set_metrics[data_set_metrics["variable_names"] == p], 
                    s=4, 
                    alpha=0.3, 
                    x="num_obs", 
                    y="values", 
                    hue="metric_name", 
                    ax=ax, 
                    hue_order=hue_order, 
                    palette=palette,
                    legend=False)
    
    ax.set_title(label, fontsize=fontsize, pad=12)
    legend = ax.legend(title="", fontsize=fontsize) 
    
    ax.set_ylabel("Metric Value", fontsize=fontsize-10)
    ax.set_xlabel("")
    ax.tick_params(axis='x', labelsize=fontsize - 17)  
    ax.tick_params(axis='y', labelsize=fontsize - 14)
    ax.set_xticks([50, 200, 400, 600, 800, 1000])

    ax.legend(loc='upper center', bbox_to_anchor=(1, -0.3),
        fancybox=False, shadow=False, ncol=3, prop={'size': 15})
    
    
    if p != model_specs['simulation_settings']['param_names'][2]:
        ax.get_legend().remove()

fig.supxlabel("Number of Observations", fontsize=fontsize-10, y=-0.1) 
fig.tight_layout()

fig.savefig(network_plot_folder + '/metrics_num_obs_' + network_name + '_random.png', bbox_inches='tight', dpi=300)
```
